Glucose-HbA1c High-Risk Patients
Find patients whose:
Glucose is above the dataset average   
HbA1c is above the dataset average   
BMI is above the dataset median  
Return the patients sorted by HbA1c descending

```python
import pandas as pd
df=pd.read_csv('Diabetes-Dataset (2).csv')

df['Glucose_mg_dL']=pd.to_numeric(df['Glucose_mg_dL'], errors='coerce')
df['HbA1c']=pd.to_numeric(df['HbA1c'], errors='coerce')
df['BMI']=pd.to_numeric(df['BMI'], errors='coerce')

glucose_avg=df['Glucose_mg_dL'].mean()
hba1c_avg=df['HbA1c'].mean()
bmi_median=df['BMI'].median()

print(f"Average Glucose:{glucose_avg:.2f}")
print(f"Average HbA1c:{hba1c_avg:.2f}")
print(f"Median BMI:{bmi_median:.2f}")

high_risk_patients=df[
    (df['Glucose_mg_dL'] > glucose_avg)&
    (df['HbA1c'] > hba1c_avg) &
    (df['BMI'] > bmi_median)

]
high_risk_patients_sorted=high_risk_patients.sort_values(by='HbA1c', ascending=False)

#display(high_risk_patients_sorted)
high_risk_patients.sort_values(by='HbA1c', ascending=False)
```

Age vs Glucose Investigation  
Calculate the average glucose for every age.  
Find:  
Age with the highest average glucose  
Age with the lowest average glucose  
Top 5 ages by average glucose  
Visualize the relationship using a line plot.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
df=pd.read_csv('Diabetes-Dataset (2).csv')

df['Glucose_mg_dL']=pd.to_numeric(df['Glucose_mg_dL'], errors='coerce')
# Convert 'Age' to numeric, coercing errors will turn non-numeric values into NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df.loc[~df['Age'].between(1, 110), 'Age'] = np.nan
df.loc[~df['Glucose_mg_dL'].between(40, 500), 'Glucose_mg_dL'] = np.nan

# Drop rows where 'Age' or 'Glucose_mg_dL' might be NaN after conversion
df_cleaned = df.dropna(subset=['Age', 'Glucose_mg_dL'])

# Calculate average glucose for every age
avg_glucose_by_age = df_cleaned.groupby('Age')['Glucose_mg_dL'].agg(['mean','count']).reset_index()
avg_glucose_by_age.columns = ['Age', 'Glucose_mg_dL', 'Count']

# Sort by average glucose in descending order
# Age with the highest average glucose
highest_glucose_age = avg_glucose_by_age.loc[avg_glucose_by_age['Glucose_mg_dL'].idxmax()]
print(f"Age with the highest average glucose: {int(highest_glucose_age['Age'])} (Average Glucose: {highest_glucose_age['Glucose_mg_dL']:.2f})")

# Age with the lowest average glucose
lowest_glucose_age = avg_glucose_by_age.loc[avg_glucose_by_age['Glucose_mg_dL'].idxmin()]
print(f"Age with the lowest average glucose: {int(lowest_glucose_age['Age'])} (Average Glucose: {lowest_glucose_age['Glucose_mg_dL']:.2f})")

# Top 5 ages by average glucose
top_5_ages_glucose = avg_glucose_by_age.sort_values(by='Glucose_mg_dL', ascending=False).head(5)
print("\nTop 5 ages by average glucose:")
display(top_5_ages_glucose)

# Visualize the relationship using a line plot
plt.figure(figsize=(14, 8))
sns.lineplot(x='Age', y='Glucose_mg_dL', data=avg_glucose_by_age, marker='o')
plt.title('Average Glucose Levels by Age')
plt.xlabel('Age')
plt.ylabel('Average Glucose (mg/dL)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45) # Rotate x-axis labels if they overlap
plt.tight_layout()
plt.show()
```

Multi-Factor Risk Score 🔥    
Create a Risk_Score:  
+1 if Glucose is above the mean  
+1 if HbA1c is above the mean  
+1 if BMI is above the median  
+1 if Age is above the median


```python
import pandas as pd
import numpy as np

df = pd.read_csv('Diabetes-Dataset (2).csv')

df['Glucose_mg_dL'] = pd.to_numeric(df['Glucose_mg_dL'], errors='coerce')
df['HbA1c'] = pd.to_numeric(df['HbA1c'], errors='coerce')
df['BMI'] = pd.to_numeric(df['BMI'], errors='coerce')
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Remove impossible values
df.loc[~df['Glucose_mg_dL'].between(40, 500), 'Glucose_mg_dL'] = np.nan
df.loc[~df['HbA1c'].between(3, 15), 'HbA1c'] = np.nan
df.loc[~df['Age'].between(1, 110), 'Age'] = np.nan

df = df.dropna(subset=['Glucose_mg_dL', 'HbA1c', 'BMI', 'Age']).copy()

glucose_avg = df['Glucose_mg_dL'].mean()
hba1c_avg = df['HbA1c'].mean()
bmi_median = df['BMI'].median()
age_median = df['Age'].median()

df['Risk_Score'] = (
    (df['Glucose_mg_dL'] > glucose_avg).astype(int) +
    (df['HbA1c'] > hba1c_avg).astype(int) +
    (df['BMI'] > bmi_median).astype(int) +
    (df['Age'] > age_median).astype(int)
)

print(df[['Glucose_mg_dL', 'HbA1c', 'BMI', 'Age', 'Risk_Score']])
```

Age–BMI Risk Segmentation
Divide patients into 4 groups based on whether their Age and BMI are above or below their respective medians:
Young + Low BMI
Young + High BMI
Older + Low BMI
Older + High BMI
For each group, calculate:
Patient count
Average Glucose
Average HbA1c
Average BMI
Determine which group has the highest average Glucose and HbA1c.


```python
import numpy as np
import pandas as pd
df=pd.read_csv('Diabetes-Dataset (2).csv')

# Convert relevant columns to numeric, coercing errors to NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['BMI'] = pd.to_numeric(df['BMI'], errors='coerce')
df['Glucose_mg_dL'] = pd.to_numeric(df['Glucose_mg_dL'], errors='coerce')
df['HbA1c'] = pd.to_numeric(df['HbA1c'], errors='coerce')

# Remove impossible values
df.loc[~df['Age'].between(1, 110), 'Age'] = np.nan
df.loc[~df['Glucose_mg_dL'].between(40, 500), 'Glucose_mg_dL'] = np.nan
df.loc[~df['HbA1c'].between(3, 15), 'HbA1c'] = np.nan

# Drop rows with NaN values in these critical columns for accurate median and group calculations
df_cleaned = df.dropna(subset=['Age', 'BMI', 'Glucose_mg_dL', 'HbA1c']).copy()

# Medians from the cleaned DataFrame
age_median = df_cleaned['Age'].median()
bmi_median = df_cleaned['BMI'].median()

# Create groups
df_cleaned['Age_Group'] = np.where(df_cleaned['Age'] >= age_median, 'Older', 'Young')
df_cleaned['BMI_Group'] = np.where(df_cleaned['BMI'] >= bmi_median, 'High BMI', 'Low BMI')
df_cleaned['Risk_Group'] = df_cleaned['Age_Group'] + ' + ' + df_cleaned['BMI_Group']

# Summary per group
summary = df_cleaned.groupby('Risk_Group').agg(
    Patient_Count=('Age', 'count'),
    Avg_Glucose=('Glucose_mg_dL', 'mean'), # Corrected column name
    Avg_HbA1c=('HbA1c', 'mean'),
    Avg_BMI=('BMI', 'mean')
).round(2)

print(summary)
print("\nHighest avg Glucose:", summary['Avg_Glucose'].idxmax())
print("Highest avg HbA1c:", summary['Avg_HbA1c'].idxmax())
```

Glucose Efficiency Analysis
Create a new column:
Glucose_HbA1c_Ratio = Glucose / HbA1c
Find:
Top 10 patients with the highest ratio
Bottom 10 patients with the lowest ratio
Average ratio by Gender
Average ratio by age group
Visualize the distribution using a histogram.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('Diabetes-Dataset (2).csv')

# Convert necessary columns to numeric, coercing errors to NaN
df['Glucose_mg_dL'] = pd.to_numeric(df['Glucose_mg_dL'], errors='coerce')
df['HbA1c'] = pd.to_numeric(df['HbA1c'], errors='coerce')
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Drop rows where Glucose_mg_dL or HbA1c are NaN to avoid issues in ratio calculation
df.dropna(subset=['Glucose_mg_dL', 'HbA1c'], inplace=True)

# New column (avoid divide-by-zero by replacing 0 with NaN in HbA1c before division)
df['Glucose_HbA1c_Ratio'] = df['Glucose_mg_dL'] / df['HbA1c'].replace(0, np.nan)

# Drop rows with NaN in Glucose_HbA1c_Ratio that might result from division by NaN or zero
df.dropna(subset=['Glucose_HbA1c_Ratio'], inplace=True)

# Top 10 and Bottom 10
print("Top 10 highest ratio:")
print(df.nlargest(10, 'Glucose_HbA1c_Ratio'))

print("\nBottom 10 lowest ratio:")
print(df.nsmallest(10, 'Glucose_HbA1c_Ratio'))

# Average ratio by Gender
print("\nAverage ratio by Gender:")
print(df.groupby('Gender')['Glucose_HbA1c_Ratio'].mean().round(2))

# Average ratio by Age group
# Drop rows where 'Age' might be NaN before creating age bands
df_age_clean = df.dropna(subset=['Age']).copy()
df_age_clean['Age_Band'] = pd.cut(df_age_clean['Age'],
                        bins=[0, 30, 45, 60, 100],
                        labels=['<30', '30-45', '45-60', '60+'])
print("\nAverage ratio by Age group:")
print(df_age_clean.groupby('Age_Band')['Glucose_HbA1c_Ratio'].mean().round(2))

# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Glucose_HbA1c_Ratio'].dropna(), bins=30, edgecolor='black')
plt.title('Distribution of Glucose / HbA1c Ratio')
plt.xlabel('Glucose_HbA1c_Ratio')
plt.ylabel('Number of Patients')
plt.show()
```
